```python
import os

from ultralytics import YOLO
from google.colab import drive, userdata
from roboflow import Roboflow
from numpy._core.defchararray import mod
```

```python
project_name = "yolov8_detectorv6"
drive.mount('/content/drive')
os.makedirs(f"/content/drive/MyDrive/{project_name}", exist_ok=True)
```

```python
!cd /content/drive/MyDrive/{project_name}
```

```python
data=userdata.get('ROBOFLOW_KEY')
os.environ["ROBOFLOW_KEY"] = data
api_key = os.environ.get("ROBOFLOW_KEY")
```

```python
os.makedirs(f"/content/drive/MyDrive/{project_name}/models/", exist_ok=True)

# Initialize Roboflow and download dataset
DATASET_VERSION = 6 # Check Roboflow for the latest version
MODEL = "yolov8"  
MODEL_VERSION = "m"  

rf = Roboflow(api_key=api_key)
project = rf.workspace("4421objectdetector").project("objectdetectionmodel-8dbp1")
dataset = project.version(DATASET_VERSION).download(MODEL, location="/content/datasets")
model = YOLO(f'yolov8{MODEL_VERSION}.pt')

results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    name='custom_detector_compost_recycling',
    device=0,
    cache=True,
    mixup=0.15,
    copy_paste=0.1,
)

```

```python
run_dir = results.save_dir  
project_dir = f"/content/drive/MyDrive/{project_name}"
best_model = f"{run_dir}/weights/best.pt"

# Verify the model file exists before copying
best_model_title = f"best_v{MODEL_VERSION}.pt"
last_model_title = f"last_v{MODEL_VERSION}.pt"

if os.path.exists(best_model):
    !cp {best_model} {project_dir}/models/{best_model_title}
    print("✅ Training complete! Model saved to Drive.")
    print(f"📁 Model location: {project_dir}/models/{best_model_title}")

    # Also copy last.pt as backup
    last_model = f"{run_dir}/weights/last.pt"
    if os.path.exists(last_model):
        !cp {last_model} {project_dir}/models/{last_model_title}
        print(f"📁 Backup model: {project_dir}/models/{last_model_title}")

print("✅ Training complete! Model saved to Drive.")
```
